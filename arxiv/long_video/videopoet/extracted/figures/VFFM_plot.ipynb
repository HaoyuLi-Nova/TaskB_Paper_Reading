```python
import matplotlib.pyplot as plt
import numpy as np

def add_value_labels(bars):
    for bar in bars:
        height = bar.get_height()
        ax.annotate('{}'.format(round(height, 2)),
                    xy=(bar.get_x() + bar.get_width() / 2, height),
                    xytext=(0, 3),  # 3 points vertical offset
                    textcoords="offset points",
                    ha='center', va='bottom',
                    fontsize=10)


fig, ax = plt.subplots(figsize=(8, 5))  # Adjust the size (width, height) as needed

# Data for plotting
model_sizes = ['300M LM\n(10B tokens)', '1B LM\n(10B tokens)', '1B LM\n(37B tokens)', '8B LM\n(37B tokens)', '8B LM\n(58B tokens)']
audio_to_video = [331.63, 283.06, 115.64, 104.09, 88.60]
frame_prediction = [65.61, 33.09, 19.56, 19.95, 12.23]
audio_video_continuation = [75.68, 41.57, 24.87, 20.04, 15.32]

# Ensure the values are in float format for the bar chart
audio_to_video = np.array(audio_to_video, dtype=float)
frame_prediction = np.array(frame_prediction, dtype=float)
audio_video_continuation = np.array(audio_video_continuation, dtype=float)

# Extend the frame_prediction array to match the others by repeating the last value
frame_prediction = np.pad(frame_prediction, (0, len(model_sizes) - len(frame_prediction)), 'edge')

# Define the width of the bars
bar_width = 0.25

# Set the positions of the bars on the x-axis
r1 = np.arange(len(audio_to_video))
r2 = [x + bar_width for x in r1]
r3 = [x + bar_width for x in r2]


# Make the bar plots
bars1 = ax.bar(r1, audio_to_video, color='#1f77b4', width=bar_width, edgecolor='grey', label='Audio to Video')
bars2 = ax.bar([x + bar_width * 2 for x in r1], frame_prediction, color='#ff7f0e', width=bar_width, edgecolor='grey', label='Frame Prediction')
bars3 = ax.bar([x + bar_width for x in r1], audio_video_continuation, color='#2ca02c', width=bar_width, edgecolor='grey', label='Audio/Video Continuation')

add_value_labels(bars1)
add_value_labels(bars2)
add_value_labels(bars3)

# Draw dashed lines at the top of the bars
for r, val in zip(r1, audio_to_video):
    ax.plot([r, r + bar_width * 2], [val, val], color='#1f77b4', linestyle='--', linewidth=1)
for r, val in zip(r2, frame_prediction):
    ax.plot([r - bar_width, r + bar_width], [val, val], color='#ff7f0e', linestyle='--', linewidth=1)
for r, val in zip(r3, audio_video_continuation):
    ax.plot([r - bar_width * 2, r], [val, val], color='#2ca02c', linestyle='--', linewidth=1)

# Add labels, title, and custom x-axis tick labels, etc.
ax.set_xlabel('Language Model Size (# Training Tokens)', fontsize=14)
ax.set_ylabel('FVD (↓) for video quality', fontsize=14)
ax.set_xticks([r + bar_width for r in range(len(audio_to_video))], fontsize=12)
ax.set_xticklabels(model_sizes)
ax.set_yscale('log')
ax.legend()

# Add grid and make layout tight
ax.grid(True, which="both", ls="--")
ax.legend(loc='upper center', bbox_to_anchor=(0.5, 1.15), ncol=3, fontsize='large')

plt.tight_layout()

# Show the plot
plt.show()
```
